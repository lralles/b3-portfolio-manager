# Rebuild the portfolio data store

This notebook removes the generated files under `data/store/` and rebuilds the complete store from the raw B3 exports using the ingestion pipeline.

Set `CONFIRM_REBUILD` to `True` in the next cell only when you want to replace the current store.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from config import Config
from src.ingestion.pipeline import run

config = Config()
STORE_DIR = config.store_dir

CONFIRM_REBUILD = True

if not CONFIRM_REBUILD:
    raise RuntimeError("Set CONFIRM_REBUILD = True before rebuilding the data store.")

for csv_path in STORE_DIR.rglob("*.csv"):
    csv_path.unlink()

STORE_DIR.mkdir(parents=True, exist_ok=True)

status = run(
    raw_transactions=config.raw_transactions_dir,
    raw_position=config.raw_position_file,
    sanitized_dir=config.sanitized_dir,
    store_dir=STORE_DIR,
    trading_prices=config.trading_prices_file,
)

status
```
