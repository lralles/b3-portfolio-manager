# Portfolio Value

This notebook loads the persisted positions, gets the latest available trading price on or before the selected date, and displays the resulting portfolio value.

```python
from datetime import date
from pathlib import Path
import sys

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(project_root))

from src.core.services.portfolio_service import PortfolioService
```

```python
portfolio_service = PortfolioService()
price_date = date.today()
portfolio = portfolio_service.get(price_date)

print(f'Price date: {price_date}')
print(f'Positions valued: {len(portfolio.position_values)}')
print(f'Portfolio value: {portfolio.portfolio_value}')
```

```python
import pandas as pd
from IPython.display import display

position_values = []
for position_value in portfolio.position_values:
    position = position_value.position
    position_values.append({
        'asset_id': position.asset_id,
        'asset_name': position.asset.asset_name if position.asset else None,
        'quantity': position.current_quantity,
        'price_date': position_value.trading_price.evaluation_date,
        'trading_price': position_value.trading_price.value,
        'position_value': position_value.value,
    })

position_value_table = pd.DataFrame(position_values)
display(position_value_table.sort_values('asset_id').reset_index(drop=True))
```
