# XIRR Portfolio Performance

This notebook calculates money-weighted portfolio performance using dated I/O transactions and the portfolio value at the end of the period. The start and end dates can be supplied below. If either is omitted, the service uses the earliest transaction date and the latest available transaction or trading-price date.

Investor cash-flow signs are used: money invested into the portfolio is negative, money withdrawn or distributed is positive, and the ending portfolio value is a positive terminal cash flow.

```python
from datetime import date
from pathlib import Path
import sys

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(project_root))

import pandas as pd
from IPython.display import display

from src.core.services import XirrPerformanceService
```

```python
# Set either parameter to None to use the service default.
START_DATE = None  # Example: date(2024, 1, 1)
END_DATE = None    # Example: date(2025, 1, 1)

result = XirrPerformanceService().calculate(
    start_date=START_DATE,
    end_date=END_DATE,
)

print(f'Period: {result.start_date} to {result.end_date}')
print(f'XIRR: {result.xirr:.2%}')
```

```python
cash_flow_table = pd.DataFrame(
    result.cash_flows,
    columns=['date', 'cash_flow'],
).assign(date=lambda table: pd.to_datetime(table['date']))

display(cash_flow_table)
```
