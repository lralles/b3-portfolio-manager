# Yearly Portfolio Performance

This notebook presents yearly portfolio performance calculated between consecutive calendar year ends. Portfolio values are evaluated on December 31; when no price exists on that day, the portfolio service uses the latest available price on or before the year end.

For year `t`, the calculation is:

`((portfolio value at the end of t - net I/O during t) / portfolio value at the end of t-1) - 1`

Net I/O is inflows minus outflows during the year.

```python
from pathlib import Path
import sys

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(project_root))

import pandas as pd
from IPython.display import display

from src.core.services import YearlyPerformanceService
```

```python
yearly_performance = YearlyPerformanceService().calculate()

performance_table = pd.DataFrame([
    {
        'year': result.year.year,
        'start_year_end': result.start_year_end,
        'end_year_end': result.end_year_end,
        'start_value': result.start_value,
        'inflow': result.inflow,
        'outflow': result.outflow,
        'net_io': result.net_io,
        'end_value': result.end_value,
        'performance': result.performance,
    }
    for result in yearly_performance
])

print(f'Years with performance: {len(performance_table)}')
display(performance_table)
```

```python
import matplotlib.pyplot as plt

chart_table = performance_table.dropna(subset=['performance'])
plt.figure(figsize=(10, 5))
plt.bar(
    chart_table['year'].astype(str),
    chart_table['performance'].astype(float) * 100,
    color=['#2a9d8f' if value >= 0 else '#e76f51' for value in chart_table['performance']],
)
plt.axhline(0, color='#333333', linewidth=0.8)
plt.title('Yearly Portfolio Performance')
plt.xlabel('Year')
plt.ylabel('Performance (%)')
plt.grid(axis='y', alpha=0.25)
plt.tight_layout()
plt.show()
```
